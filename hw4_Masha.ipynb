```python
import random as rnd
import string as s

alphabet = s.ascii_letters

def generate_password(length):
    new_password = ""
    
    
    for i in range(length):
        random_letter = rnd.choice(alphabet)
        random_num = rnd.randint(0, 9)
        random_symbol = rnd.choice("!@$%^&*.")
        new_password += random_letter + str(random_num) + random_symbol
    return new_password[0:length]
    
    


generate_password(10)
```

количество слов
уникальные слова колво
самое часто встречающееся и сколько раз
среднюю длину слова

```python
from collections import Counter as c

st = input("Введите строку")
split_st = st.split()

def word_num(st):
    return len(split_st)

def word_uni(st):
    st.lower()
    return len(set(split_st))

def word_freq_num(st):
    return c(split_st).most_common(1)
        

def word_average(st):
    len_word = 0
    for i in split_st:
       len_word += len(i) 

    return len_word / len(split_st)
    
print(word_num(st))
print(word_uni(st))
print(word_freq_num(st))
print(word_average(st))
```

Напишите программу «Таймер». Пользователь вводит количество секунд. Программа отсчитывает время, каждую секунду выводя оставшееся время в формате MM:SS. По истечении времени выводится сообщение и проигрывается звуковой сигнал (print('\a')).

```python
import time
import datetime

def timer():
    seconds = int(input("Введите колво секунд"))

    for i in range(seconds, -1, -1):
        print(datetime.time(0, 0, i).strftime("%M:%S"))
        time.sleep(1)
    print('\a')

timer() 
```

Напишите функцию calculate(expression), которая принимает строку вида "3 + 5 * 2" (поддерживаются +, -, *, /, ^ для степени) и возвращает результат с учётом приоритета операций. Используйте обратную польскую запись или рекурсивный спуск. eval использовать нельзя.

```python
def calculate(expression):
    parts = expression.split()
    numbers = []
    operators = []
    priority = {'+': 1, '-': 1, '*': 2, '/': 2, '^': 3}
    
    for i in range(len(parts)):
        if i % 2 == 0:
            numbers.append(int(parts[i]))
        else:
            operators.append(parts[i])
    
    i = 0
    while i < len(operators):
        if operators[i] in '^*/':
            if operators[i] == '^':
                numbers[i] = numbers[i] ** numbers[i + 1]
            elif operators[i] == '*':
                numbers[i] = numbers[i] * numbers[i + 1]
            elif operators[i] == '/':
                numbers[i] = numbers[i] / numbers[i + 1]
            
            numbers.pop(i + 1)
            operators.pop(i)
            i -= 1
        i += 1
    
    result = numbers[0]
    for i in range(len(operators)):
        if operators[i] == '+':
            result += numbers[i + 1]
        else:
            result -= numbers[i + 1]
    
    return int(result)

print(calculate("3 + 5 * 2"))
```
