```python
#1
def equation_quadratic(coeff_1, coeff_2, constant):
    discriminant = coeff_2 ** 2 - 4 * coeff_1 * constant
    x1 = (-coeff_2 + discriminant ** 0.5) / (2 * coeff_1)
    x2 = (-coeff_2 - discriminant ** 0.5) / (2 * coeff_1)
    print(x1, x2, end=' ')
    return x1, x2
    
def equation_biquadratic(coeff_1, coeff_2, constant):
    discriminant = coeff_2 ** 2 - 4 * coeff_1 * constant
    y1 = (-coeff_2 + discriminant ** 0.5) / (2 * coeff_1)
    y2 = (-coeff_2 - discriminant ** 0.5) / (2 * coeff_1)
    print(y1 ** 0.5, -y1 ** 0.5, y2 ** 0.5, -y2 ** 0.5)


equation_quadratic(1, -12, 20)
print()
equation_quadratic(1, 17, 72)
print()
equation_quadratic(1, -7, -44)
print()
equation_quadratic(1, 9, 8)
print()
equation_quadratic(1, -2, -63)
print()
equation_biquadratic(1, -12, 27)
```

```python
#2
def reverse(my_list):
    reversed_list = []
    i = len(my_list) - 1
    
    while i >= 0:
        reversed_list.append(my_list[i])
        i -= 1
        
    return reversed_list

reverse([1, 2, 3, 4])
```

```python
#3
def change(lst):
    temp = lst[0]
    lst[0] = lst[len(lst)-1]
    lst[len(lst)-1] = temp
    print(lst)

my_list = [1, 2, 3, 4, 5]

change(my_list)
```

```python
#4
def useless(s):
    print(max(s) / len(s))

useless([1, 3, 5, 20])
```

```python
#5
def all_eq(lst):
    new_list = []
    max_len = 0
    
    for word in lst:
        if len(word) > max_len:
            max_len = len(word) 
            
    for word in lst:
        if len(word) < max_len:
            word += (max_len - len(word)) * '_'
        new_list.append(word)
            
    return new_list
    
print(all_eq(['крот', 'белка', 'выхухоль']))
```
