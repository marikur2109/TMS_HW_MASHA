```python
#1
import math

x = 1.79

def result(x):
    y = (math.cos(math.exp(x))
         + math.log((1 + x) ** 2)
         + math.sqrt(math.exp(math.cos(x))
         + (math.sin(math.pi * x)) ** 2)
         + math.sqrt(1 / x)
         + math.cos(x ** 2)) ** math.sin(x)
    return y
    
print(result(x))
```

```python
#2
def search_substr(substr, st):
    substr = substr.lower()
    st = st.lower()
    
    if (substr in st): 
        print("Есть контакт!")
    else:
        print("Мимо!")
        
search_substr("привет", "ПрИВет мир!")
```

```python
#3
import collections

def count_freq(st):
    st = st.replace(" ", "")
    print(collections.Counter(st).most_common(3))
    
count_freq("Эта книга адресована всем, кто изучает русский язык.")
```

```python
#4
import collections

def count_it(sequence):
    return dict(collections.Counter(sequence).most_common(3))
    
print(count_it("2111111444444555555522"))        
```

```python
#5.1
def is_indanger_q(a, b, c, d):
    
    if (a == c) or (b == d) or (abs(a-c) == abs(b-d)):
        print("Ферзь угрожает")
    else:
        print("Ферзь не угрожает")
        
is_indanger_q(1, 2, 2, 1)
```

```python
#5.2
def is_indanger_k(a, b, c, d):
    
    if ((a == c-3 or a == c+3) and (b == d-1 or b == d+1)) or ((a == c-1 or a == c+1) and (b == d-3 or b == d+3)):
        print("Конь угрожает")
    else:
        print("Конь не угрожает")
        
is_indanger_k(3, 2, 0, 1)
```
