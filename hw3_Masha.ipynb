```python
#3
def list_int(num):
    new_list = []
    
    for i in range(num, 0, -1):
        new_list.append(i)
        
    print(new_list)

list_int(5)
```

```python
#4
def decade(year):
    
    if year % 100 != 0:
        print(int(year / 100 + 1))
    else:
        print(int(year / 100))

decade(1705)
decade(1900)
decade(1601)
```

```python
#5
def reverse_num(num):
    str_num = str(num)
    new_list = []
    
    for i in range(len(str_num)-1, -1, -1):
        new_list.append(str_num[i])
        
    print(new_list)

reverse_num(35231)
```

```python
#6
def vertical_matrix(n):
    my_matrix = []
    num = 1
    
    for i in range(n):
        my_matrix.append([0] * n)
        
    for col in range(n):
        if col % 2 == 0:
            for row in range(n):
                my_matrix[row][col] = num
                num += 1
        
        else:
            for row in range(n-1, -1, -1):
                my_matrix[row][col] = num
                num += 1

    for row in my_matrix:
        print(row)

def horizontal_matrix(n):
    my_matrix = []
    num = 1
    
    for i in range(n):
        my_matrix.append([0] * n)
        
    for row in range(n):
        if row % 2 == 0:
            for col in range(n):
                my_matrix[row][col] = num
                num += 1
        else:
            for col in range(n-1, -1, -1):
                my_matrix[row][col] = num
                num += 1

    for col in my_matrix:
        print(col)
        
vertical_matrix(3)
print()
horizontal_matrix(3)
```
